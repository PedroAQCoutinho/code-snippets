```python
import requests
import xml.etree.ElementTree as ET
```

```python
url = "https://geoserver.car.gov.br/geoserver/sicar/ows?service=WFS&version=2.0.0&request=GetCapabilities"

# Faz a requisição ignorando o certificado SSL
response = requests.get(url, verify=False)

# Parseando a resposta XML
root = ET.fromstring(response.content)
```

```python
# Encontrar todas as camadas disponíveis
namespaces = {'wfs': 'http://www.opengis.net/wfs/2.0'}
layers = root.findall(".//{http://www.opengis.net/wfs/2.0}FeatureType")

print("Camadas disponíveis no GeoServer:")
for layer in layers:
    name = layer.find("{http://www.opengis.net/wfs/2.0}Name").text
    title = layer.find("{http://www.opengis.net/wfs/2.0}Title")
    title_text = title.text if title is not None else "Sem título"
    print(f"- {name}: {title_text}")
```
